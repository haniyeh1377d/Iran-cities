# 10 Major Cities of Iran
Population (2026 estimates), geographic coordinates, descriptions, and charts.

**Data note:** Iran's last official census was in 2016, so the 2026 figures are estimates (source: World Population Review projections based on census series). The 2016 census figures are included for comparison. All charts use the 2026 estimate.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cities = [
    # City, Title, Population 2026 (estimate), Population 2016 (census), Latitude, Longitude, Description
    ("Tehran",     "Capital of Iran",           9882527, 8693706, 35.6892, 51.3890, "Largest city and capital of Iran; the political, economic and cultural center of the country."),
    ("Mashhad",    "The Holy City",             3532584, 3001184, 36.2605, 59.6168, "Capital of Razavi Khorasan Province; home to the shrine of Imam Reza, a major pilgrimage site."),
    ("Isfahan",    "Half of the World",         2196430, 1961260, 32.6546, 51.6680, "Historic city famous for Naqsh-e Jahan Square, Persian architecture and handicrafts."),
    ("Karaj",      "Gateway to the Alborz",     1549130, 1592492, 35.8400, 50.9391, "Capital of Alborz Province, located next to Tehran; a major industrial and residential city."),
    ("Shiraz",     "City of Poets",             1798531, 1565572, 29.5918, 52.5837, "Capital of Fars Province; city of Hafez and Saadi, close to the ruins of Persepolis."),
    ("Tabriz",     "Historic Trade Hub",        1694340, 1558693, 38.0962, 46.2738, "Capital of East Azerbaijan; known for its UNESCO-listed historic bazaar and heavy industry."),
    ("Qom",        "Religious Center",          1502321, 1201158, 34.6416, 50.8746, "Major center of Shia religious study; home to the shrine of Fatima Masumeh."),
    ("Ahvaz",      "City of Bridges",           1344918, 1184788, 31.3183, 48.6706, "Capital of Khuzestan Province on the Karun River; a hub for oil and agriculture."),
    ("Kermanshah", "Gateway to the West",       1170300, 946651,  34.3142, 47.0650, "Capital of Kermanshah Province; known for ancient sites such as Taq-e Bostan and Bisotun."),
    ("Urmia",      "City by the Lake",          895630, 736224,  37.5527, 45.0761, "Capital of West Azerbaijan Province, located near Lake Urmia; known for its orchards and grapes."),
]
df = pd.DataFrame(cities, columns=["City", "Title", "Population", "Population_2016", "Latitude", "Longitude", "Description"])
pd.set_option("display.max_colwidth", None)
df
```

## Pie Chart

```python
plt.figure(figsize=(8, 8))
plt.pie(df["Population"], labels=df["City"], autopct="%1.1f%%", startangle=90)
plt.title("Population Share of 10 Major Iranian Cities")
plt.show()
```

## Bar Chart

```python
plt.figure(figsize=(10, 5))
plt.bar(df["City"], df["Population"], color="steelblue")
plt.xticks(rotation=45)
plt.xlabel("City")
plt.ylabel("Population")
plt.title("Population of 10 Major Iranian Cities")
plt.tight_layout()
plt.show()
```

## Scatter Plot
City locations by longitude and latitude. Marker size is proportional to population.

```python
plt.figure(figsize=(9, 7))
plt.scatter(df["Longitude"], df["Latitude"],
            s=df["Population"] / 8000, alpha=0.6, c=df["Population"], cmap="viridis")
for i, r in df.iterrows():
    plt.annotate(r["City"], (r["Longitude"], r["Latitude"]),
                 textcoords="offset points", xytext=(6, 6))
plt.colorbar(label="Population")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Location of Cities (marker size = population)")
plt.grid(alpha=0.3)
plt.show()
```

## Line Chart
Cities sorted by population.

```python
d = df.sort_values("Population", ascending=False)
plt.figure(figsize=(10, 5))
plt.plot(d["City"], d["Population"], marker="o", color="crimson")
plt.xticks(rotation=45)
plt.xlabel("City")
plt.ylabel("Population")
plt.title("Cities Sorted by Population")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
